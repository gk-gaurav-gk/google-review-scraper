# Text Representation and Vectorization in NLP

## Objective

After cleaning and preprocessing text, we must convert it into a numerical form that machine learning models can understand.  
This process is known as **text vectorization**.  

In this notebook, we will:
- Understand the need for numerical representation of text.
- Learn about two key vectorization methods:
  1. Bag of Words (BoW)
  2. TF–IDF (Term Frequency–Inverse Document Frequency)
- Implement both using scikit-learn.
- Compare their outputs and interpretation.


````python
# Step 1: Import Libraries

from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
import pandas as pd
````

## Step 2: Why Text Representation?

Machine learning models cannot understand words — they only process **numbers**.  
Hence, every text document must be represented as a numeric vector.  
The goal is to preserve the **semantic meaning** of the text while making it computable.

Let’s take a small collection of text documents, called a **corpus**.


````python
# Step 2: Create a simple text corpus

corpus = [
    "Natural Language Processing is fascinating",
    "I love studying NLP and Machine Learning",
    "Deep Learning improves NLP performance",
    "NLP is part of Artificial Intelligence"
]

print("Sample text corpus:")
for i, doc in enumerate(corpus, 1):
    print(f"{i}. {doc}")
````

## Step 3: Bag of Words (BoW)

### Concept
Bag of Words represents text by counting how many times each word appears.  
It ignores grammar, order, and context — treating the text as a “bag” of individual words.

### Example
If we have two sentences:
- “I love NLP”
- “NLP is fun”

Then the vocabulary = [I, love, NLP, is, fun]

| Sentence | I | love | NLP | is | fun |
|-----------|---|------|-----|----|-----|
| I love NLP | 1 | 1 | 1 | 0 | 0 |
| NLP is fun | 0 | 0 | 1 | 1 | 1 |

Now, let’s implement this idea using `CountVectorizer`.


````python
# Step 3: Implement Bag of Words

vectorizer = CountVectorizer()
X_bow = vectorizer.fit_transform(corpus)

# Convert to DataFrame for better visualization
bow_df = pd.DataFrame(X_bow.toarray(), columns=vectorizer.get_feature_names_out())

print("Bag of Words Representation:\n")
print(bow_df)
````

## Step 4: Term Frequency–Inverse Document Frequency (TF–IDF)

### Concept
BoW treats all words equally.  
However, some words (like *is*, *the*, *and*) appear often but carry little meaning.  
TF–IDF solves this by giving higher weight to **unique words** and lower weight to **common words**.

It uses:
- **TF (Term Frequency):** How frequently a word appears in a document.  
- **IDF (Inverse Document Frequency):** How rare that word is across all documents.

The result is a **weighted representation** showing which words are truly important.


````python
# Step 4: Implement TF–IDF

tfidf_vectorizer = TfidfVectorizer()
X_tfidf = tfidf_vectorizer.fit_transform(corpus)

# Convert to DataFrame
tfidf_df = pd.DataFrame(X_tfidf.toarray(), columns=tfidf_vectorizer.get_feature_names_out())

print("TF–IDF Representation:\n")
print(tfidf_df.round(3))  # rounded for readability
````

## Step 5: Compare Bag of Words and TF–IDF

| Feature | Bag of Words | TF–IDF |
|----------|---------------|--------|
| Meaning | Counts word occurrences | Weighs words by importance |
| Common Words | All treated equally | Common words get lower weight |
| Rarity Handling | Ignores rarity | Highlights rare but meaningful words |
| Output Type | Integer counts | Floating-point weights |

TF–IDF is preferred for most NLP applications because it highlights informative words while downplaying unimportant ones.


## Step 6: Beyond TF–IDF — Word Embeddings

Bag of Words and TF–IDF treat words as independent features, ignoring relationships and context.  
To overcome this, NLP uses **word embeddings** — dense numerical vectors that represent semantic meaning.

### What are Word Embeddings?
Word embeddings map words to continuous vector spaces where **similar words lie close together**.  
This allows models to understand relationships such as gender, tense, or semantic similarity.

### Popular Word Embedding Methods
- **Word2Vec (Google)** – Learns vectors using neural networks (CBOW and Skip-Gram).
- **GloVe (Stanford)** – Builds embeddings from global word co-occurrence statistics.
- **FastText (Meta)** – Uses subword information (character n-grams) to handle rare words.
- **BERT / Transformer-based models** – Generate context-dependent embeddings.

### Why Use Embeddings?
- Capture both **syntactic** and **semantic** relationships.  
- Help models generalize better.  
- Reduce the curse of dimensionality compared to Bag-of-Words or TF–IDF.
